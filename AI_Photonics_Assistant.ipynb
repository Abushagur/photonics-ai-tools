```python
def analyze_fabry_perot(name, R, n, d_um, lambda_nm=1550, verbose=True):
    """
    Complete AI-powered Fabry-Perot cavity analysis.
    
    Parameters:
    name      : cavity name
    R         : mirror reflectivity (0 to 1)
    n         : refractive index of cavity medium
    d_um      : cavity length in microns
    lambda_nm : operating wavelength in nm
    """
    # Step 1 — compute parameters
    finesse   = np.pi * np.sqrt(R) / (1 - R)
    lambda_m  = lambda_nm * 1e-9
    d_m       = d_um * 1e-6
    FSR_nm    = (lambda_m**2 / (2 * n * d_m)) * 1e9
    linewidth = FSR_nm / finesse
    
    # Step 2 — ask Claude
    question = f"""
    Analyze this Fabry-Perot optical cavity:
    
    Cavity name: {name}
    Mirror reflectivity R: {R}
    Cavity length: {d_um} um
    Refractive index: {n}
    Operating wavelength: {lambda_nm} nm
    
    Computed parameters:
    - Finesse: {finesse:.2f}
    - Free spectral range (FSR): {FSR_nm:.4f} nm
    - Transmission linewidth (FWHM): {linewidth:.4f} nm
    
    Please provide:
    1. Suitability as a narrowband optical filter
    2. Suitability as a laser cavity
    3. Key tradeoffs and limitations
    4. Recommended improvements
    
    Keep each point to 2-3 sentences. Be specific and technical.
    """
    
    try:
        message = client.messages.create(
            model="claude-sonnet-4-6",
            max_tokens=1024,
            messages=[{"role": "user", "content": question}]
        )
        claude_analysis = message.content[0].text
    except Exception as e:
        claude_analysis = f"API call failed: {e}"
    
    results = {
        "name":      name,
        "finesse":   finesse,
        "FSR_nm":    FSR_nm,
        "linewidth": linewidth,
        "claude_analysis": claude_analysis
    }
    
    if verbose:
        print("=" * 60)
        print(f"FABRY-PEROT ANALYSIS REPORT — {name}")
        print("=" * 60)
        print(f"Mirror reflectivity: {R:.2f}")
        print(f"Cavity length:       {d_um:.2f} um")
        print(f"Finesse:             {finesse:.2f}")
        print(f"FSR:                 {FSR_nm:.4f} nm")
        print(f"Linewidth:           {linewidth:.4f} nm")
        print()
        print("Claude AI Engineering Analysis:")
        print("-" * 60)
        print(claude_analysis)
        print("=" * 60)
    
    return results
```

```python
def analyze_fiber(name, core_um, n1, n2, lambda_nm=1550, verbose=True):
    # Step 1 — compute all parameters
    NA = np.sqrt(n1**2 - n2**2)
    V  = (2 * np.pi * core_um*1e-6 / (lambda_nm*1e-9)) * NA
    acceptance_angle = np.degrees(np.arcsin(NA))
    
    # Find cutoff wavelength
    lambdas = np.linspace(500, 2000, 10000)
    V_curve = (2 * np.pi * core_um*1e-6 / (lambdas*1e-9)) * NA
    cutoff_idx = np.argmin(np.abs(V_curve - 2.405))
    cutoff_lambda = lambdas[cutoff_idx]
    
    # Mode classification
    if V < 2.405:
        mode = "single-mode"
    elif V < 3.832:
        mode = "few-mode"
    else:
        mode = "multimode"
    
    # Step 2 — ask Claude
    question = f"""
    Analyze this step-index optical fiber:
    Fiber name: {name}
    Core radius: {core_um} um
    n1: {n1}, n2: {n2}
    Wavelength: {lambda_nm} nm
    NA: {NA:.4f}, V-number: {V:.4f}
    Mode: {mode}
    Cutoff wavelength: {cutoff_lambda:.1f} nm
    
    Provide:
    1. Suitability for single-mode operation
    2. Strengths for DWDM
    3. Main limitations
    4. Recommended applications
    Keep each point to 2-3 sentences.
    """
    
    try:
        message = client.messages.create(
            model="claude-sonnet-4-6",
            max_tokens=1024,
            messages=[{"role": "user", "content": question}]
        )
        claude_analysis = message.content[0].text
    except Exception as e:
        claude_analysis = f"API call failed: {e}"
    
    # Step 3 — display
    if verbose:
        print("=" * 60)
        print(f"FIBER ANALYSIS REPORT — {name}")
        print("=" * 60)
        print(f"Core diameter:       {core_um*2:.1f} um")
        print(f"NA:                  {NA:.4f}")
        print(f"V-number:            {V:.4f}")
        print(f"Mode:                {mode}")
        print(f"Acceptance angle:    {acceptance_angle:.2f} degrees")
        print(f"Cutoff wavelength:   {cutoff_lambda:.1f} nm")
        print()
        print("Claude AI Engineering Analysis:")
        print("-" * 60)
        print(claude_analysis)
        print("=" * 60)
    
    return {"name": name, "NA": NA, "V_number": V, 
            "mode": mode, "claude_analysis": claude_analysis}
```
